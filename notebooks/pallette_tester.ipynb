```python
import numpy as np
import matplotlib.pyplot as plt

import apoda as ap
from apoda import style

schema = ap.BOTANICA      # Select schema
style.apply(schema)     # Apply style (normal palette)
```

```python
x = np.linspace(0, 10, 200)

plt.figure()
plt.plot(x, np.sin(x), label="sin")
plt.plot(x, np.cos(x), label="cos")

plt.title("Line Plot (Default Cycle)")
plt.legend()
plt.savefig("line_plot_default_cycle.png", dpi=300)
plt.show()
```

```python
style.apply(schema, palette="sequential", n=6)

plt.figure()

for i in range(6):
    y = np.sin(x + i * 0.3)
    plt.plot(x, y)

plt.title("Sequential Palette")
plt.show()
```

```python
x_rand = np.random.rand(200)
y_rand = np.random.rand(200)
y_rand2 = np.random.rand(200)

style.apply(schema)

plt.figure()
plt.scatter(x_rand, y_rand, alpha=0.7)
plt.scatter(x_rand, y_rand2, alpha=0.7)
plt.title("Scatter")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
style.apply(schema, palette="ordinal")

data = np.random.randn(1000)
data2 = np.random.randn(1000)
data3 = np.random.randn(1000)

plt.figure()
plt.hist(data, bins=30, alpha=0.9, label="Original")
plt.hist(data2+1, bins=30, alpha=0.9, label="Shifted")

plt.hist(data3+2, bins=30, alpha=0.9, label="Shifted")
plt.title("Histogram")
plt.legend()
plt.show()
```

```python
style.apply(schema)

fig, ax = plt.subplots()

ax.plot(x, np.sin(x))

style.helper_hline(ax, 0.5, schema)

ax.set_title("Helper Line Test")
ap.sparkle()
plt.show()
```

```python
style.apply(
    schema,
    palette="ordinal",
)

plt.figure()

for i in range(10):
    plt.plot(x, np.sin(x + i * 0.2))

plt.title("Color Cycle Stress Test")
ap.sparkle(color=["white", "#7dd3fc", "#c084fc"])
plt.show()
```

```python
style.apply(schema)

plt.figure()
for i in range(10):
    plt.plot(x, np.sin(x)*i)

plt.title("Grid & Background Check")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```
